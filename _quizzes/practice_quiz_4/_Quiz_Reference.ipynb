## 4. Selecting by Position with `.iloc`

**What / when:** Use this when the question says "use `.iloc`", gives row numbers ("cars 28–32"), or asks for column **positions** instead of names.

`.iloc[rows, columns]`, where `start:stop` **includes start and excludes stop**.

**Source:** Lecture 8, exercises 3–6; Quiz 4 C.

````python
# All rows, a range of columns
<df_subset> = <df>.iloc[:, <start_col>:<stop_col>]

# A range of rows AND a range of columns
<df_subset> = <df>.iloc[<start_row>:<stop_row>, <start_col>:<stop_col>]

# "All the way to the end" (leave stop blank)
<df_subset> = <df>.iloc[<start_row>:<stop_row>, <start_col>:]

# First n rows, all columns
<df_subset> = <df>.iloc[:<n>, :]

# Specific (non-consecutive) rows: use a list
<df_subset> = <df>.iloc[[0, 1, 23, 24], :]

# One column as a Series, by position
<series> = <df>.iloc[:, <col_position>]

# One row as a Series, by position
<row> = <df>.iloc[<row_position>, :]
````

**Course examples**
- `carfeatures.iloc[:, 2:5]` gives displacement, horsepower and weight (columns 2, 3, 4).
- `carfeatures.iloc[28:33, 4:]` gives cars 28–32 with weight, acceleration and vehicle_id (Lecture 8 and Quiz 4 C).
- `df_ht_wt.iloc[:, 0]` gives the height column as a Series (Lecture 8).

**Common mistakes**
- **The stop number is excluded.** Cars 28 through 32 is `28:33`, not `28:32`.
- Counting starts at **0**. The 5th column is position 4.
- The question says "use `:` for both row and column indices", so write `28:33, 4:7` or `28:33, 4:` and **not** a list like `[28, 29, 30, 31, 32]`.
- `.iloc` uses **positions**, not names. `df.iloc[:, "weight"]` is an error.
- Separate rows and columns with a **comma** inside one pair of brackets: `.iloc[a:b, c:d]`.

## 6. Filtering Rows with `.query`

**What / when:** Use this to keep rows that meet a condition ("cars with exactly 8 cylinders and mpg ≥ 25", "players taller than 175").

The condition is written **as a string**.

**Source:** Lecture 8 (query, `and`, `@var` in a loop); Quiz 4 A (`<`, `>=`); Quiz 4 F (backticks).

````python
# One condition
<df_subset> = <df>.query("<col> > <value>")

# Two conditions
<df_subset> = <df>.query("<col1> == <value1> and <col2> >= <value2>")

# Save the condition string in a variable first (also fine)
my_expression = "<col1> == <value1> and <col2> >= <value2>"
<df_subset> = <df>.query(my_expression)

# Use a PYTHON VARIABLE inside the query with @
<threshold> = <value>
<df_subset> = <df>.query("<col> == @<threshold>")

# Column name has spaces / parentheses: wrap it in BACKTICKS ` `
<df_subset> = <df>.query(" `<Col Name (units)>` > <value> ")
````

**Course examples**
- `carfeatures.query("cylinders == 8 and mpg >= 25")` (Lecture 8)
- `df_fifa.query("random_var < 0.4")` (Quiz 4 A)
- ``df_fifa.query(" `Height(in cm)` > 175 ")`` (Quiz 4 F)
- `carfeatures.query("cylinders == @cyl")` inside a `for` loop (Lecture 8)

**Comparison operators:** `==` equal · `!=` not equal · `>` · `>=` · `<` · `<=` · combine with `and` / `or`

**Common mistakes**
- `=` assigns and `==` compares. A query needs `==`.
- "**Strictly** greater than" is `>`, and "at least" or "≥" is `>=`.
- **Backticks** `` ` `` (the key left of 1) are not single quotes `'`. Use backticks for column names with spaces or parentheses.
- Forgetting `@` before a Python variable gives an "undefined name" error.
- *(General pandas, not shown in the course files:)* comparing to text needs inner quotes: `df.query("group == 'control'")`.

## 12. Merging with `pd.merge`

**What / when:** Two DataFrames share an ID column (`title`, `raceId`, `state`, `constructorId`) and you want the info from one added onto the other.

| Argument | Meaning |
|---|---|
| `left` | the **primary** DataFrame (its columns come first) |
| `right` | the **secondary** DataFrame (the info being added) |
| `how` | `"left"` keeps every row of `left` in **left's** order · `"right"` keeps every row of `right` in **right's** order |
| `on` | the shared column used to match rows (a string, or a list of strings) |

**Source:** Lecture 9a; Quiz 4 B and E; Assignment 8 (c).

````python
# BASIC: left is primary, add info from right, match on a shared column
<df_merged> = pd.merge(left = <df_primary>, right = <df_secondary>, how = "left", on = "<key_col>")

# ONLY MERGE CERTAIN COLUMNS: subset the right DataFrame first with [[...]]
#   -> ALWAYS include the key column in the subset!
right_df = <df_secondary>[["<key_col>", "<col_to_add1>", "<col_to_add2>"]]
<df_merged> = pd.merge(left = <df_primary>, right = right_df, how = "left", on = "<key_col>")

# Same thing in one line (Lecture 9a's last cell)
<df_merged> = pd.merge(left = <df_primary>,
                       right = <df_secondary>[["<key_col>", "<col_to_add>"]],
                       how = "left", on = "<key_col>")

display(<df_merged>)
````

**Course examples**
- Lecture 9a: `pd.merge(left = results_short, right = races_short, how = "left", on = "raceId")`
- HW8 (c): `pd.merge(left = df_movies, right = df_scores[["title", "imdb_score"]], how = "left", on = "title")`
- Quiz 4 E: `right_df = df_const[["constructorId", "name", "nationality"]]`, then `pd.merge(left=df_sprint, right=right_df, on="constructorId", how="left")`

**Common mistakes**
- **Leaving the key column out of the subset.** `df[["imdb_score"]]` has no `"title"`, so the merge fails. Use `df[["title", "imdb_score"]]`.
- Using `.drop` after merging when the question says to merge only the required columns (Quiz 4 E forbids this). Subset **before** merging.
- `on` must be a column name that exists in **both** DataFrames, spelled the same.
- A `NaN` in the merged result means that key had no match in the other DataFrame.

## 13. Merging: Controlling Row Order and Column Order

**What / when:** The question says "columns in this order: ..." or "the ordering of the rows should match file X".

**Rules (from Lecture 9a):**
- **Column order:** `left`'s columns come first, then `right`'s new columns.
- **Row order:** `how = "left"` follows `left`'s row order, and `how = "right"` follows `right`'s row order.

So you can pick **which columns go first** with `left`/`right`, and **whose row order wins** with `how`, separately.

| You want... | Do this |
|---|---|
| A's columns first **and** A's row order | `left = A, right = B, how = "left"` |
| B's columns first **but** A's row order | `left = B, right = A, how = "right"` ← **Quiz 4 B trick** |

````python
# Quiz 4 B pattern: the result must START with columns from the small table,
# but keep the ROW ORDER of the big table.

left_df  = <df_small>[["<col_first>", "<key_col>"]]                          # columns you want first
right_df = <df_big>[["<key_col>", "<col_a>", "<col_b>", "<col_c>"]]          # rows you want kept in order

<sol> = pd.merge(left = left_df, right = right_df, how = "right", on = "<key_col>")
<sol>.head()
````

**Course example (Quiz 4 B):** the target columns were `"region", "state", "county", "candidate", "total_votes"` in the CSV's row order.
```python
left_df = df_region[["region", "state"]]
right_df = df_pres_county[["state", "county", "candidate", "total_votes"]]
sol = pd.merge(left=left_df, right=right_df, how="right", on="state")
```
- `[["region", "state"]]` puts region **before** state.
- `how="right"` keeps the CSV's row order.

**Lecture 9a exercise:** try all 4 combinations (`left`/`right` swapped × `how="left"`/`"right"`) and compare the outputs to see the rules.

**Common mistakes**
- The order **inside** `[[...]]` sets the column order. `[["state", "region"]]` would put state first.
- Swapping `left` and `right` without changing `how` changes the row order too.

## 14. Merging: Duplicate Column Names, Multiple Keys, Chained Merges

**Source:** Lecture 9a (F1 results → races → circuits; the DOB/Name example).

````python
# (a) Both tables have a column with the SAME NAME that is NOT the key
#     -> pandas renames them <name>_x and <name>_y. Use suffixes to choose better names:
<df_merged> = pd.merge(left = <df_left>, right = <df_right>, how = "left", on = "<key_col>",
                       suffixes = ("_<left_tag>", "_<right_tag>"))

# (b) Match on TWO columns at once (rows must agree on both)
<df_merged> = pd.merge(left = <df_left>, right = <df_right>, how = "left", on = ["<key1>", "<key2>"])

# (c) Three tables: merge twice (output of the first merge becomes the next left)
<df_1> = pd.merge(left = <df_A>[["<keyAB>", "<colA>"]],
                  right = <df_B>[["<keyAB>", "<keyBC>", "<colB>"]], how = "left", on = "<keyAB>")
<df_2> = pd.merge(left = <df_1>,
                  right = <df_C>[["<keyBC>", "<colC>"]], how = "left", on = "<keyBC>")
````

## 15. Cleaning Messy Values

**What / when:** A column that should hold numbers contains text like `"1,250"`, `"~60"`, `"40 mg"`, `"(25)"`, `"eighteen"`. Find them, fix them, then convert the column to numbers.

**Source:** Assignment 8 (a).

````python
# Part 1: find the unique values that are NOT plain digits
numeric_bool = <df>["<col>"].str.isnumeric()               # True if only digits
non_numeric_vals = <df>["<col>"][numeric_bool == False]   # keep the False ones
unique_vals = pd.unique(non_numeric_vals)                 # drop repeats
print(unique_vals)

# Part 2: map each messy value to a clean one, then convert to numbers
replace_dict = {"<messy1>": "<clean1>", "<messy2>": "<clean2>"}   # built from Part 1's output

<df>["<col>"] = <df>["<col>"].replace(replace_dict)
<df>["<col>"] = pd.to_numeric(<df>["<col>"])

display(<df>)
````

## 17. Linear Regression with `smf.ols` and Plotting the Fit

**What / when:** "Fit a linear model y ≈ a·x + b", "compute the coefficients", "plot the data and the best fit line".

- `a` = slope, labeled with the x column's name (e.g. `"weight"`)
- `b` = y-intercept, always labeled `"Intercept"` (capital I)

**Source:** Assignment 8 (d) and (e).

````python
import statsmodels.formula.api as smf

<df> = pd.read_csv("<folder>/<file>.csv")

# "dependent ~ independent"  (y ~ x)
model = smf.ols(formula = "<y_col> ~ <x_col>", data = <df>)
coeffs = model.fit().params          # Series: Intercept, <x_col>
print(coeffs)

# Pull out slope and intercept
a = coeffs["<x_col>"]                # slope
b = coeffs["Intercept"]              # intercept (capital I!)

# Predicted y for every row (vectorized, like NumPy)
predicted_<y> = a * <df>["<x_col>"] + b

# Plot: data = scatter, model = line, on the SAME figure
plt.scatter(x = <df>["<x_col>"], y = <df>["<y_col>"])
plt.plot(<df>["<x_col>"], predicted_<y>, color = "red")

plt.legend(labels = ["data", "best fit line"])
plt.xlabel("<x label>")
plt.ylabel("<y label>")
plt.title("<title>")
plt.show()
````

**HW8 example:** `smf.ols(formula = "mpg ~ weight", data = carfeatures)`, with `a = coeffs["weight"]` and `b = coeffs["Intercept"]`.

**Common mistakes**
- The formula is **one string** with a tilde `~`, and y goes on the **left**.
- `coeffs["intercept"]` (lowercase) is a KeyError. It's `"Intercept"`.
- Mixing up `a` and `b`: the slope is the x column's coefficient.
- Use `plt.plot` for the line (the question says "**not** a scatter plot").
- Calling `plt.show()` between the scatter and the line gives two separate figures.
